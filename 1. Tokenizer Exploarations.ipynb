# Python Regular Expression Tokenization

This code **splits raw text (`raw_text`) into tokens (words, punctuation marks, etc.) and removes unnecessary whitespace.**

## 1. Import the Regular Expression Library

```python
import re
```

- `re` is Python's built-in Regular Expression library.
- It is used for pattern matching, splitting, and replacing text.

## 2. Split Text Using Regular Expressions

```python
preprocessed = re.split(r"([,.:;?_!\"()']|--|\s)", raw_text)
```

`re.split()` splits text based on specified regular expression patterns.

| Regular Expression | Description |
|---|---|
| `[,.:;?_!"()']` | Matches specified punctuation marks |
| `--` | Matches two consecutive hyphens |
| `\s` | Matches whitespace characters (spaces, newlines, tabs, etc.) |
| `\|` | Logical OR, matching any of the specified patterns |
| `()` | Capturing group that preserves delimiters in the output |

**Key Point:**

Normally, `re.split()` removes delimiters from the resulting list. However, using a capturing group `()` preserves them, allowing punctuation marks to become separate tokens.

## 3. Remove Unnecessary Whitespace

```python
preprocessed = [item.strip() for item in preprocessed if item.strip()]
```

This is a **List Comprehension**.

- `item.strip()`: Removes leading and trailing whitespace.
- `if item.strip()`: Filters out empty strings and whitespace-only strings.

## 4. Display the First 30 Tokens

```python
print(preprocessed[:30])
```

- `[:30]` retrieves the first 30 elements of the list.
- Python uses zero-based indexing.

In LLM, the model will:
- Firstly **endode** the tokens into token IDs:
    
    splitting text into token, carrying out the `string-to-integer` mapping to produce token IDs into the dictionary which are going to be our vocabulary list.
- Then **decode** the tokens from their token IDs:
    
    Carrying out the `integer-to-string` mapping to produce token IDs back to text.

We are going to create a class for this, where we already have an intuition of what the class should include: **Encoder** and **Decoder**.
## 3. Coding a tokenizer class in Python

Python Regex: Removing Whitespace Before Punctuation

## 1. Code Example

```python
import re

text = "Hello , world ! How are you ?"

text = re.sub(r'\s+([,.?!"()])', r'\1', text)

print(text)
# Output: Hello, world! How are you?
```

**Purpose:** Remove whitespace before punctuation marks while preserving the punctuation.

## 2. Regex Breakdown

```python
re.sub(r'\s+([,.?!"()])', r'\1', text)
```

| Component | Meaning |
|---|---|
| `re.sub()` | Replaces text matching a regex pattern |
| `r'...'` | Raw string that preserves backslashes |
| `\s+` | Matches one or more whitespace characters |
| `[...]` | Character class matching one character from the set |
| `,.?!"()` | Punctuation characters to match |
| `(...)` | Capturing group 1 |
| `\1` | Reuses the text captured by group 1 |
| `text` | Original string being processed |

## 3. How It Works

### Step 1: Match Whitespace and Punctuation

Input:

```text
Hello , world !
```

Regex matches:

```text
" ," → Space + comma
" !" → Space + exclamation mark
```

### Step 2: Capture Punctuation

The parentheses `(...)` capture the punctuation.

```text
Group 1 = ","
Group 1 = "!"
```

### Step 3: Replace Using `\1`

The replacement `r'\1'` keeps only the captured punctuation.

```text
" ," → ","
" !" → "!"
```

Output:

```text
Hello, world!
```

## 4. Including Apostrophes

To include both single and double quotation marks, use a triple-quoted raw string:

```python
import re

text = "Hello , world ! I 'm learning Python ."

text = re.sub(r"""\s+([,.?!"()'])""", r'\1', text)

print(text)
# Output: Hello, world! I'm learning Python.
```

## 5. Key Takeaways

- `\s+` → Matches one or more whitespace characters.
- `[...]` → Matches one character from the specified set.
- `(...)` → Captures matched text into a group.
- `\1` → References the first captured group.
- `re.sub()` → Replaces matched text with the specified replacement.
- `r'...'` → Prevents Python from interpreting backslashes as escape sequences.

**Important:** `\1` does not replace only the whitespace. It replaces the **entire regex match** with the punctuation captured in group 1.

### Summary

```python
re.sub(r'\s+([,.?!"()])', r'\1', text)
```

**Result:** Removes whitespace before punctuation while keeping the punctuation unchanged.

````python
class SimpleTokenizerV1:
    def __init__(self, vocab):
        # Store the vocabulary as a class attribute for access in the encode and decode methods
        self.str_to_int = vocab # ENCODING
        # Note that vocab = {token:integer for integer, token in enumerate(all_words)}
        # Create an INVERSE vocabulary that maps token IDs back to the original text tokens
        self.int_to_str = {i:s for s, i in vocab.items()} # DECODING
    
    #Process input text into token IDs
    def encode(self, text):
        # Splits input text into tokens using regex (punctuation + whitespace as separators)
        preprocessed = re.split(r'([,.:;?_!"()\']|--|\s)', text)
        # Cleans tokens with strip() and filters out empty strings.
        preprocessed = [item.strip() for item in preprocessed if item.strip()]
        ids = [self.str_to_int[s] for s in preprocessed]
        return ids
    #Convert token IDs back into tokens
    def decode(self, ids):
        text = " ".join([self.int_to_str[i] for i in ids])
        #Joins them with spaces.
        text = re.sub(r'\s+([,.?!"()\'])', r'\1', text)
        #Uses re.sub to remove unwanted spaces before punctuation (so you don’t get "Hello !", but "Hello!").
        return text
````

Create an instance and try out!

````python
tokenizer = SimpleTokenizerV1(vocab)

text = """"It's the last he painted, you know," 
           Mrs. Gisburn said with pardonable pride."""
#Notice this text snippet is directly copy pasted from the story
ids = tokenizer.encode(text)
print(ids)
````

## 4. Improving the tokenizer class
As we can see the current tokenizer is unable to encode words that is absent in the vocabulary.

Let's modify the tokenizer to handle unknown words! But, how?

You might have guessed it! We were talking about tokens all along, so why not add special tokens?

We can modify the tokenizer to use an `<|unk|>` token if it encounters a word that is not part of the vocabulary. 

Also, we are adding a token `<|endoftext|>` token between unrelated texts. In fact, when training GPT-like LLMs on multiple independent documents or books, it is common to insert a token before each document or book that follows a previous text source.

Let's start with extending the token list!

````python
all_tokens = sorted(list(set(preprocessed)))
all_tokens.extend(["<|endoftext|>", "<|unk|>"])

vocab = {token:index for index, token in enumerate(all_tokens)}

len(vocab.items())
for i, item in enumerate(list(vocab.items())[-5:]):
    print(item)
````

Hmmm, exactly 2 more tokens than before. Good.
And the appended tokens are right there!

Let's improve the code that we have already written and try out!

````python
class SimpleTokenizerV2:
    def __init__(self, vocab):
        self.str_to_int = vocab
        self.int_to_str = {i:s for s,i in vocab.items()}
    
    def encode(self, text):
        preprocessed = re.split(r'([,.:;?_!"()\']|--|\s)', text)                       
        preprocessed = [item.strip() for item in preprocessed if item.strip()]
        preprocessed = [
            item if item in self.str_to_int
            else "<|unk|>" for item in preprocessed
        ]
        # Replace unknown words by <|unk|> tokens
        ids = [self.str_to_int[s] for s in preprocessed]
        return ids
        
    def decode(self, ids):
        text = " ".join([self.int_to_str[i] for i in ids])
        # Replace spaces before the specified punctuations
        text = re.sub(r'\s+([,.?!"()\'])', r'\1', text)
        return text
````

````python
tokenizer = SimpleTokenizerV2(vocab)

text1 = "Hello, do you like tea?"
text2 = "In the sunlit terraces of the palace"

text = " <|endoftext|> ".join((text1, text2))

print(text)
````

````python
tokenizer.encode(text)
````

````python
tokenizer.decode(tokenizer.encode(text))
````

### Fun fact
The tokenizer of GPT model only uses an `<|endoftext|>` token for simplicity, it also doesn't use an `<|unk|>` token for out-of-vocabulary words. 

Instead, GPT models use a **Byte Pair Encoding (BPE)** tokenizer, breaking down words into subword units, which we will cover just a bit later! This is why GPT model will never encounter a out-of-vocabulary scenario.
<div class='alert alert-block alert-success'>
As we can see, those <|unk|> tokens are those words that are not appeared in the text, while the <|endoftext|> token is the seperator between two seperated sentences.
</div>

## 5. Byte Pair Encoding (BPE)

Let me first tell you a truth: the BPE tokenizer was used to train LLMs such as GPT-2, 3, 4, 5 and 6 that we use everyday☠️

But what on earth is BPE?
<div class ='alert alert-block alert-info'>
BPE is a sub-word tokenization algorithm that starts with individual bytes/characters and repeatedly merges the most frequently occuring adjacent pairs of tokens into new single tokens -- creating a vocabulary that contains both whole words and common subword fragments.
</div>

But this time, we are not going to implement from scratch! Instead we are going to use an existing Python open-source library called [`tiktoken`](https://github.com/openai/tiktoken). 

This library implements the BPE algorithm very efficiently based on source code in Rust.

Once installed, we can instantiate and try out the BPE tokenizer from `tiktoken` as follows:

````python
tokenizer = tiktoken.get_encoding("gpt2")
# It's that easy
````

````python
text = (
    "Hello, do you like tea? <|endoftext|> In the sunlit terraces"
     "of someunknownPlace."
)

integers = tokenizer.encode(text, allowed_special={"<|endoftext|>"})

print(integers)

strings = tokenizer.decode(integers)

print(strings)
````

We can make two noteworthy observations based on the token IDs and decoded text
above. 

<div class='alert alert-block alert-info'>
First, the <|endoftext|> token is assigned a relatively large token ID, namely,
50256. 
</div>

In fact, the BPE tokenizer, which was used to train models such as GPT-2, GPT-3,
and the original model (GPT-3.5) used in ChatGPT, has a total vocabulary size of 50,257, with
<|endoftext|> being assigned the largest token ID.

<div class='alert alert-block alert-info'>
Second, the BPE tokenizer above encodes and decodes unknown words, such as
"someunknownPlace" correctly. 

The BPE tokenizer can handle any unknown word. How does
it achieve this without using <|unk|> tokens?
</div>

Let us take another simple example to illustrate how the BPE tokenizer deals with `unknown tokens`:

````python
integers = tokenizer.encode("Akwirw ier")
print(integers)

strings = tokenizer.decode(integers)
print(strings)
````

<div class="alert alert-block alert-warning">
The algorithm underlying BPE breaks down words that aren't in its predefined vocabulary
into smaller subword units or even individual characters.

The enables it to handle out-of-vocabulary words. 

So, thanks to the BPE algorithm, if the tokenizer encounters an
unfamiliar word during tokenization, it can represent it as a sequence of subword tokens or
characters.
</div>

````python
import tiktoken

# Initialize the encodings for GPT-2, GPT-3, and GPT-4
encodings = {
    "GPT-2": tiktoken.get_encoding("gpt2"),
    "GPT-3": tiktoken.get_encoding("p50k_base"),
    "GPT-3.5": tiktoken.get_encoding("cl100k_base"),
    "GPT-4": tiktoken.get_encoding("cl100k_base"),
    "GPT-4o": tiktoken.get_encoding("o200k_base"),
    "GPT-4.1": tiktoken.get_encoding("o200k_base"),
    "GPT-5": tiktoken.get_encoding("o200k_base"),
    "GPT-OSS": tiktoken.get_encoding("o200k_harmony")
}

# Get the vocabulary size for each encoding
vocab_sizes = {model: encoding.n_vocab for model, encoding in encodings.items()}

# Print the vocabulary sizes
for model, size in vocab_sizes.items():
    print(f"The vocabulary size for {model} is: {size}")
````
